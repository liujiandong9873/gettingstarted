```python
import numpy as np

import matplotlib.pyplot as plt

t = np.linspace(0, 2 * np.pi, 1000)
x = 16 * np.sin(t)**3
y = 13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)

plt.plot(x, y, color='red')
plt.title('Heart Shape')
plt.axis('equal')
plt.show()
# Create an "ugly" heart shape by adding some noise to the y values
noise = np.random.normal(0, 0.5, y.shape)
y_ugly = y + noise

plt.plot(x, y_ugly, color='blue')
plt.title('Ugly Heart Shape')
plt.axis('equal')
plt.show()
```
